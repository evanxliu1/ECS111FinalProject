# 00 — Setup: connect the repo to Google Drive

**Run this first, once per Colab session.** It mounts Drive and syncs the GitHub repo into your Drive project folder.

Layout this produces — code and data live together at `/content/drive/MyDrive/ECS111FinalProject`:
- **code** (`notebooks/`, `src/`, `configs/`) — tracked by git, source of truth is GitHub
- **data** (`data/`, `checkpoints/`) — gitignored, persists in Drive between sessions

Workflow: edit and `git push` code from your laptop → re-run this notebook in Colab to `git pull` the latest. Data you generate in Colab stays in Drive and is never committed.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os

REPO_URL = 'https://github.com/evanxliu1/ECS111FinalProject.git'
BRANCH = 'main'
PROJECT_DIR = '/content/drive/MyDrive/ECS111FinalProject'
DRIVE_ROOT = '/content/drive/MyDrive'

if os.path.isdir(os.path.join(PROJECT_DIR, '.git')):
    # Already a clone — just pull the latest code. Gitignored data is untouched.
    print('repo already present, pulling latest...')
    !cd "{PROJECT_DIR}" && git pull origin {BRANCH}
elif os.path.isdir(PROJECT_DIR):
    # Folder exists (e.g. you already generated data here) but isn't a git repo yet.
    # Attach git in-place: init, add remote, fetch, hard-checkout the code.
    # Safe for data files — they're gitignored, so checkout won't touch them.
    print('folder exists without git — attaching repo in-place...')
    !cd "{PROJECT_DIR}" && git init -q && git remote add origin {REPO_URL} \
        && git fetch -q origin {BRANCH} && git checkout -f -b {BRANCH} origin/{BRANCH}
else:
    # Fresh clone straight into Drive.
    print('cloning fresh into Drive...')
    !cd "{DRIVE_ROOT}" && git clone {REPO_URL} ECS111FinalProject

%cd $PROJECT_DIR
```

```python
# Verify: show repo state and what data is already present in Drive
!git log --oneline -3
print()
for sub in ['data/raw', 'data/generated', 'data/splits', 'checkpoints']:
    p = os.path.join(PROJECT_DIR, sub)
    items = os.listdir(p) if os.path.isdir(p) else []
    print(f'{sub}: {items if items else "(empty)"}')
```
